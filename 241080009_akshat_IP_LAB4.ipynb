```python
import cv2
import matplotlib.pyplot as plt
from skimage.filters import roberts, prewitt, sobel,laplace
from skimage.feature import canny
```

```python
#og image
img=cv2.imread("ip2.jfif")
img=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
#grayscale
gray=cv2.cvtColor(img,cv2.COLOR_RGB2GRAY)
```

```python
#no gaussian blur
#roberts
roberts_edges=roberts(gray)

#prewitt
prewitt_edges=prewitt(gray)

#sobel
sobel_edges=sobel(gray)

#laplacian
laplacian_edges=laplace(gray)

#canny
canny_edges=canny(gray)

plt.figure(figsize=(40,10))
plt.subplot(1,7,1)
plt.imshow(img)
plt.title("og")
plt.axis("off")


plt.subplot(1,7,2)
plt.imshow(gray,cmap="gray")
plt.title("gray")
plt.axis("off")

plt.subplot(1,7,3)
plt.imshow(roberts_edges,cmap="gray")
plt.title("robert")
plt.axis("off")

plt.subplot(1,7,4)
plt.imshow(prewitt_edges,cmap="gray")
plt.title("prewitt")
plt.axis("off")

plt.subplot(1,7,5)
plt.imshow(sobel_edges,cmap="gray")
plt.title("sobel")
plt.axis("off")

plt.subplot(1,7,6)
plt.imshow(laplacian_edges,cmap="gray")
plt.title("laplacian")
plt.axis("off")

plt.subplot(1,7,7)
plt.imshow(canny_edges,cmap="gray")
plt.title("canny")
plt.axis("off")
plt.show()
```

```python
#after applying gausian blur
blur = cv2.GaussianBlur(gray, (5, 5), 0)

#roberts
roberts_edges=roberts(blur)

#prewitt
prewitt_edges=prewitt(blur)

#sobel
sobel_edges=sobel(blur)

#laplacian
laplacian_edges=laplace(blur)

#canny
canny_edges=canny(blur)

plt.figure(figsize=(40,10))
plt.subplot(1,7,1)
plt.imshow(img)
plt.title("og")
plt.axis("off")


plt.subplot(1,7,2)
plt.imshow(gray,cmap="gray")
plt.title("gray")
plt.axis("off")

plt.subplot(1,7,3)
plt.imshow(roberts_edges,cmap="gray")
plt.title("robert")
plt.axis("off")

plt.subplot(1,7,4)
plt.imshow(prewitt_edges,cmap="gray")
plt.title("prewitt")
plt.axis("off")

plt.subplot(1,7,5)
plt.imshow(sobel_edges,cmap="gray")
plt.title("sobel")
plt.axis("off")

plt.subplot(1,7,6)
plt.imshow(laplacian_edges,cmap="gray")
plt.title("laplacian")
plt.axis("off")

plt.subplot(1,7,7)
plt.imshow(canny_edges,cmap="gray")
plt.title("canny")
plt.axis("off")
plt.show()
```
