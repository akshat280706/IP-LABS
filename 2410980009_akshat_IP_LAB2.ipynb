```python
import cv2
import matplotlib.pyplot as plt
img=cv2.imread("ip2_2.jpeg")
img_rgb=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
mean_filtered=cv2.blur(img_rgb,(3,3))
plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(img_rgb)
plt.title("og image")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(mean_filtered)
plt.title("mean filtered image")
plt.axis("off")


plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt
img=cv2.imread("ip2_2.jpeg")
img_rgb=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
gaussian_filtered=cv2.GaussianBlur(img_rgb,(5,5),0)
plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(img_rgb)
plt.title("og image")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(gaussian_filtered)
plt.title("gaussian filtered image")
plt.axis("off")


plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt
img=cv2.imread("ip2_2.jpeg")
img_rgb=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
median_filtered=cv2.medianBlur(img_rgb,5)
plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(img_rgb)
plt.title("og image")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(median_filtered)
plt.title("median filtered image")
plt.axis("off")


plt.show()
```
