```python
!pip install scikit-image -q
import cv2
import numpy as np
import matplotlib.pyplot as plt
from skimage.exposure import match_histograms
from google.colab import files
```

```python
print("original grayscale image: ")
uploaded = files.upload()

filename = list(uploaded.keys())[0]

original = cv2.imread(filename, cv2.IMREAD_GRAYSCALE)

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(original, cmap='gray')
plt.title("og grayscale img: ")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.hist(original.ravel(), bins=256, range=[0, 256])
plt.title("Original Histogram")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")

plt.show()
```

```python
equalized = cv2.equalizeHist(original)

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(equalized, cmap='gray')
plt.title("Histogram Equalized Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.hist(equalized.ravel(), bins=256, range=[0, 256])
plt.title("Equalized Histogram")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")

plt.show()
```

```python
print("refrence grayscale img:")
uploaded = files.upload()

filename_ref = list(uploaded.keys())[0]
reference = cv2.imread(filename_ref, cv2.IMREAD_GRAYSCALE)
```

```python
matched = match_histograms(original, reference)

matched = matched.astype(np.uint8)
```

```python
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(original, cmap='gray')
plt.title("og img: ")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(reference, cmap='gray')
plt.title("refrence img: ")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(equalized, cmap='gray')
plt.title("equlized img:")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(matched, cmap='gray')
plt.title("Histogram Matched Image")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python

plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.hist(original.ravel(), bins=256, range=[0, 256])
plt.title("Original Histogram")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")

plt.subplot(2, 2, 2)
plt.hist(equalized.ravel(), bins=256, range=[0, 256])
plt.title("Equalized Histogram")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")

plt.subplot(2, 2, 3)
plt.hist(reference.ravel(), bins=256, range=[0, 256])
plt.title("Reference Histogram")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")

plt.subplot(2, 2, 4)
plt.hist(matched.ravel(), bins=256, range=[0, 256])
plt.title("Matched Histogram")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```
